```python
import pandas as pd
import plotly.express as px

air_pollution_sensors_data = pd.read_csv("air_pollution_sensors_data.csv")
locations = pd.read_csv("locations.csv")
taxi_orders = pd.read_csv("taxi_orders.csv")

air_pollution_sensors_data.head(), locations.head(), taxi_orders.head()
```

```python
air_pollution_sensors_data['date'] = pd.to_datetime(air_pollution_sensors_data['date'])

avg_pollution = air_pollution_sensors_data.groupby('location_id')[['pm2_5concmass1hourmean_value', 'pm10concmass1hourmean_value']].mean().reset_index()

location_pollution = locations.merge(avg_pollution, left_on='id', right_on='location_id')

location_pollution.rename(columns={
    'pm2_5concmass1hourmean_value': 'avg_pm25',
    'pm10concmass1hourmean_value': 'avg_pm10',
    'name': 'Station Name'
}, inplace=True)

fig_pm25 = px.scatter_map(location_pollution,
                          lat="latitude", lon="longitude",
                          size="avg_pm25",
                          color="avg_pm25",
                          hover_name="Station Name",
                          color_continuous_scale=px.colors.sequential.Bluered,
                          size_max=50,
                          zoom=12,
                          map_style="carto-positron",
                          title="Average PM2.5 Levels by Station")

fig_pm10 = px.scatter_map(location_pollution,
                          lat="latitude", lon="longitude",
                          size="avg_pm10",
                          color="avg_pm10",
                          hover_name="Station Name",
                          color_continuous_scale=px.colors.sequential.Bluered,
                          size_max=50,
                          zoom=12,
                          map_style="carto-positron",
                          title="Average PM10 Levels by Station")



taxi_rides = taxi_orders.groupby(['originLng', 'oroginLat']).size().reset_index(name='ride_count')

fig_taxi_rides = px.scatter_map(taxi_rides,
                                   lat="oroginLat", lon="originLng",
                                   size="ride_count",
                                   color="ride_count",
                                   hover_name="ride_count",
                                   color_continuous_scale=px.colors.sequential.Oryel_r,
                                   size_max=10,
                                   zoom=12,
                                   map_style="carto-positron",
                                   title="Taxi Rides by Location")

fig_pm25.show(renderer='browser')
fig_pm10.show(renderer='browser')
fig_taxi_rides.show(renderer='browser')
# if browser rendering does not work just do .show()
```
